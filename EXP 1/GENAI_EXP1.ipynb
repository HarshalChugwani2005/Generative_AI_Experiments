## Step 4: Import Required Modules

```python
from transformers import pipeline

print("Transformers library imported successfully.")
```

## Step 5: Create a Text-Generation Pipeline

```python
# Load GPT-2 text generation model
generator = pipeline("text-generation", model="gpt2")
print("GPT-2 text-generation pipeline created successfully.")
```

## Step 6: Execute a Sample Generation

```python
output = generator(
    "Generative AI is transforming education because",
    max_length=60,
    num_return_sequences=1
)
print(output[0]["generated_text"])
```

## Step 7: Modify the Prompt — Repeat with Five Different Prompts

```python
prompts = [
    "Artificial Intelligence will shape the future of",
    "The most important skill for a data scientist is",
    "In the next ten years, robots will",
    "Climate change can be addressed by",
    "The best way to learn programming is"
]

results = {}

for p in prompts:
    output = generator(
        p,
        max_length=60,
        num_return_sequences=1,
        do_sample=True,
        top_k=50
    )
    generated = output[0]["generated_text"]
    results[p] = generated
    print("Prompt:", p)
    print("Generated:", generated)
    print("-" * 80)
```

## Step 8: Experiment with Generation Parameters

Vary `max_length`, `temperature`, `do_sample`, and `top_k` to observe how the generated text changes.

```python
prompt = "Artificial Intelligence will shape the future of"

param_sets = [
    {"max_length": 40,  "temperature": 0.3, "do_sample": True, "top_k": 50},
    {"max_length": 60,  "temperature": 0.8, "do_sample": True, "top_k": 50},
    {"max_length": 80,  "temperature": 1.2, "do_sample": True, "top_k": 50},
    {"max_length": 60,  "temperature": 0.8, "do_sample": True, "top_k": 10},
    {"max_length": 60,  "temperature": 0.8, "do_sample": False, "top_k": 50},
]

for i, params in enumerate(param_sets, 1):
    result = generator(prompt, num_return_sequences=1, **params)
    print(f"Configuration {i}: {params}")
    print("Generated:", result[0]["generated_text"])
    print("-" * 80)
```
